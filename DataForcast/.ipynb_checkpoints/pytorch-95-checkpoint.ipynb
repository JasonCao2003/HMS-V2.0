```python
import pymysql

# 连接MySQL数据库
connection = pymysql.connect(
    host='localhost',
    user='root',
    password='123456',
    database='hms'
)

try:
    with connection.cursor() as cursor:
        # 查询某月份某星期的数据
        query = '''
            SELECT 
                YEAR(orderdate) AS year,
                MONTH(orderdate) AS month,
                WEEK(orderdate) AS week,
                COUNT(*) AS count
            FROM 
                orders
            WHERE 
                YEAR(orderdate) = 2023 AND MONTH(orderdate) = 11
            GROUP BY 
                YEAR(orderdate), MONTH(orderdate), WEEK(orderdate)
            ORDER BY 
                YEAR(orderdate), MONTH(orderdate), WEEK(orderdate)
        '''

        # 执行查询
        cursor.execute(query)

        # 获取查询结果
        result = cursor.fetchall()

        # 构建结果列表
        data = []
        for row in result:
            year = row[0]
            month = row[1]
            week = row[2]
            count = row[3]
            data.append({
                'year': year,
                'month': month,
                'week': week,
                'count': count
            })

        # 打印结果
        for item in data:
            print(item)

finally:
    # 关闭数据库连接
    connection.close()
```

```python
import pymysql
import datetime

# 连接MySQL数据库
conn = pymysql.connect(host='localhost', 
                       user='root', 
                       password='123456',
                       database='hms',
                       charset='utf8mb4')
                           
# 通过Cursor执行SQL语句        
cursor = conn.cursor()

# 指定统计月份
month = 11 

# 统计每个星期几的订单数量
sql = "SELECT DAYOFWEEK(orderdate) day, COUNT(*) count FROM orders WHERE MONTH(orderdate)={} GROUP BY day".format(month)

cursor.execute(sql)

# 获取结果集
results = cursor.fetchall()

# 打印结果 
for result in results:
    day = ''
    if result[0] == 1:
        day = '星期一'
    elif result[0] == 2:
        day = '星期二'
    elif result[0] == 3:
        day = '星期三' 
    elif result[0] == 4:
        day = '星期四'
    elif result[0] == 5:
        day = '星期五'
    elif result[0] == 6:
        day = '星期六'
    elif result[0] == 7:  
        day = '星期天'
        
    print('{}有{}条数据'.format(day, result[1]))
        
# 关闭连接
cursor.close()
conn.close()
```

```python
import pymysql
import csv

conn = pymysql.connect(host='localhost', 
                       user='root', 
                       password='123456',
                       database='hms')
                       
cursor = conn.cursor()

# 构建SQL查询所有月份和星期的订单条数       
sql = """
SELECT MONTH(orderdate) month, DAYOFWEEK(orderdate) day, COUNT(*) count  
FROM orders
GROUP BY month, day
"""

cursor.execute(sql)
results = cursor.fetchall()

# 将结果写入CSV文件 
fieldnames = ['month', 'day', 'count']
with open('train_data.csv', 'w') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows([{'month':month, 'day':day, 'count':count} for month,day,count in results])
    
print(result)
cursor.close()
conn.close()
```
